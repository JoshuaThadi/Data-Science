# Exploratory Data Analysis on Olympic

```python
# import necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the dataset
data = pd.read_csv('C:\\Users\\Joshua\\Downloads\\EDA\\Olympic_EDA\\dataset_olympics.csv')
```

```python
# display the first few rows of the dataset
data.head()
```

```python
# display dataset information
data.info()
```

```python
# discribe the dataset
data.describe()
```

```python
# describe categorical columns
data.describe(include='object')
```

```python
# check for missing values
data.isna().sum()
```

```python
# check for duplicate rows
data.duplicated().sum()
```

```python
# remove duplicate rows
data.drop_duplicates(inplace=True)
```

```python
data.duplicated().sum()
```

```python
# visualize the distribution of athletes
sns.countplot(data=data, x='Sex')
plt.title('Distribution by Sex')
plt.show()
```

```python
# visualize the distribution of ages
sns.histplot(data=data, x='Age', bins=20, kde=True)
plt.title('Age Distribution of Atheletes')
plt.show()
```

```python
# visualize the distribution of heights
sns.histplot(data=data, x='Height', bins=20, kde=True)
plt.title('Height Distribution of Athletes')
plt.show()
```

```python
# visualize the distribution of heights
sns.histplot(data=data, x='Weight', bins=20, kde=True)
plt.title('Weight Distribution of Athletes')
plt.show()
```

```python
# visualize the distribution of medals
sns.countplot(data=data, x='Medal')
plt.title('Medal Distribution')
plt.show()
```

```python
# visualize medal distribution over the years
sns.countplot(data=data, x="Year", hue="Medal", palette="Set2")
plt.title("Medal Distribution Over the Years")
plt.show()
```

```python
# find the average age of athletes for each year
year_avg_age = data.groupby('Year')['Age'].mean()
print(year_avg_age)
```

```python
# find the sports with median height
sport_median_height = data.groupby('Sport')['Height'].median()
print(sport_median_height)
```

```python
# find the sports with the highest median heights
print(sport_median_height.sort_values(ascending=False).head(10))
```

```python
# find the sports with the lowest median heights
print(sport_median_height.sort_values(ascending=True).head(10))
```

```python
# find the maximum and minimum median heights across sports
print(sport_median_height.max()); print(sport_median_height.min())
```

```python
#  find the sports with median height of 170 cm
sport_median_height[sport_median_height == 170.0]
```

```python
# count of athletes gender-wise for each country
country_gender_count = data.groupby(['NOC', 'Sex'])["ID"].count()
print(country_gender_count)
```

```python
# find the number of gold medals won by each country
country_gold_medals = data[data["Medal"] == "Gold"].groupby("NOC")["Medal"].count()
print(country_gold_medals)
```

```python
# find the countries with exactly 747 gold medals
country_gold_medals[country_gold_medals == 747]
```

```python
# find the countries with more than 100 gold medals
country_gold_medals[country_gold_medals > 100]
```

```python
# find the countries with more than 1000 gold medals
if (country_gold_medals[country_gold_medals > 1000]).empty:
    print("None")
```

```python
# Find the maximum and minimum gold medals won by a country
print(country_gold_medals.max()); print(country_gold_medals.min())
```

```python
# Total medals won by each country
country_medals = data[data["Medal"].notna()].groupby("NOC")["Medal"].count()
print(country_medals)
```

```python
# Average weight of athletes by sport
sport_gender_avg_weight = data.groupby(["Sport", "Sex"])["Weight"].mean()
print(sport_gender_avg_weight)
```

```python
# Number of unique events in each sport
sport_event_count = data.groupby("Sport")["Event"].nunique().sort_values(ascending=False);
sport_event_count.plot(kind='bar', figsize=(12,6), title='Number of Unique Events per Sport');
plt.xlabel("Sport")
plt.ylabel("Number of Unique Events")
plt.xticks(rotation=45)
plt.show()
```

```python
# Number of unique athletes participating each year
year_participation = data.groupby("Year")["ID"].nunique()
year_participation.plot(kind="line", figsize=(12,6), marker='o')
plt.title("Number of Unique Athletes Participating Each Year")
plt.xlabel("Year")
plt.ylabel("Number of Unique Athletes")
plt.show()
```

```python
# Top 10 Countries with highest Average Athlete Age of participants
country_avg_age = data.groupby("NOC")["Age"].mean().sort_values(ascending=False)
country_avg_age.head(10).plot(kind='bar', figsize=(12,6), title='Top 10 Countries with highest Average Athlete Age of participants')
plt.xlabel("Country (NOC)")
plt.ylabel("Average Age")
plt.xticks(rotation=45)
plt.show()
```

```python
# Top 10 Countries with Lowest Average Athlete Age of participants
country_avg_age = data.groupby("NOC")["Age"].mean().sort_values(ascending=False)
country_avg_age.tail(10).plot(kind='bar', figsize=(12,6), title='Top 10 Countries with lowest Average Athlete Age of participants')
plt.xlabel("Country (NOC)")
plt.ylabel("Average Age")
plt.xticks(rotation=45)
plt.show()
```

```python
# age distribution by season
sns.boxplot(data=data, x="Season", y="Age")
plt.title("Age Distribution by Season")
plt.xlabel("Season")
plt.ylabel("Age")
plt.show()
```

```python
# height distribution by medal type
sns.violinplot(data=data, x="Medal", y="Height", palette="Set3")
plt.title("Height Distribution by Medal Type")
plt.xlabel("Medal")
plt.ylabel("Height")
plt.show()
```

```python
# find the country with most medals
most_medals_country = data['NOC'].value_counts().idxmax()
print(f"Country with most medals: {most_medals_country}")
```

```python
# Find the tallest athlete
tallest_athlete = data[data['Height'] == data["Height"].max()]
print("Tallest Athlete Detail:")
print(tallest_athlete[["ID", "Name", "Height", "Sport", "Year"]])
```

```python
# Find the shortest athlete
shortest_athlete = data[data['Height'] == data["Height"].min()]
print("Shortest Athlete Details:")
print(shortest_athlete[["ID", "Name", "Height", "Sport", "Year"]])
```

```python
# Find the heaviest athlete
heaviest_athlete = data[data['Weight'] == data["Weight"].max()]
print("Heaviest Athlete Detail:")
print(heaviest_athlete[["ID", "Name", "Weight", "Sport", "Year"]])
```

```python
# Find the lightest athlete
lightest_athlete = data[data['Weight'] == data["Weight"].min()]
print("Lightest Athlete Details:")
print(lightest_athlete[["ID", "Name", "Weight", "Sport", "Year"]])
```

```python
# Height vs Weight scatter plot colored by Medal type
sns.scatterplot(data=data, x='Height', y='Weight', hue="Medal", palette="Set2")
plt.title("Height vs Weight of Athletes by Medal Type")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.legend(title = "Medal")
plt.show()
```

```python
# Heatmap of medals won by each country over the years
medals_by_country_year = data.pivot_table(index="City", columns="Year", values="Medal", aggfunc='count')
sns.heatmap(medals_by_country_year, annot=True, fmt=".0f", cmap="YlGnBu")
```
